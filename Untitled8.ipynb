```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python

# Read the image using OpenCV
image = cv2.imread('highway lane.jpg')

if image is None:
    raise FileNotFoundError("Image not found. Keep building.jpg in the same folder as this notebook.")

original_image = image.copy()
```

```python

# Step 3: Convert the image to grayscale
gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
```

```python

# Input image and grayscale image
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))  # Convert image to RGB for displaying
plt.title("Input Image")
plt.axis('off')
```

```python
plt.imshow(gray_image, cmap='gray')
plt.title("Grayscale Image")
plt.axis('off')
```

```python
# Step 4: Using Canny operator from cv2, detect the edges of the image
edges = cv2.Canny(gray_image, 50, 150)  # Canny edge detection with threshold values 50 and 150
```

```python
# Apply binary thresholding
_, threshold = cv2.threshold(gray_image, 127, 255, cv2.THRESH_BINARY)
```

```python

plt.figure(figsize=(8, 10))
plt.imshow(threshold, cmap='gray')
plt.title("Thresholded Image")
plt.axis('off')
plt.show()
```

```python

height, width = threshold.shape

# Define a broad polygonal region of interest
roi_vertices = np.array([[
    (0, height),
    (0, int(height * 0.45)),
    (width, int(height * 0.45)),
    (width, height)
]], dtype=np.int32)

roi_mask = np.zeros_like(threshold)
cv2.fillPoly(roi_mask, roi_vertices, 255)
roi_image = cv2.bitwise_and(threshold, roi_mask)
```

```python

plt.figure(figsize=(8, 10))
plt.imshow(roi_image, cmap='gray')
plt.title("ROI Masked Image")
plt.axis('off')
plt.show()
```

```python

# Canny Edge Detector output
plt.imshow(edges, cmap='gray')
plt.title("Canny Edge Detector")
plt.axis('off')
```

```python

blurred = cv2.GaussianBlur(edges, (5, 5), 0)
```

```python

plt.figure(figsize=(8, 10))
plt.imshow(blurred, cmap='gray')
plt.title("Smoothed Image")
plt.axis('off')
plt.show()
```

```python

# Step 5: Detect lines using HoughLinesP
lines = cv2.HoughLinesP(
    edges,
    1,
    np.pi / 180,
    100,
    minLineLength=50,
    maxLineGap=10
)
```

```python
for line in lines:
    x1, y1, x2, y2 = line[0]  # Access the 0th element to unpack the 4 values
    cv2.line(image, (x1, y1), (x2, y2), (0, 255, 0), 2)
```

```python

# Display the result of Hough Transform (Image with lines)
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))  # Image with lines drawn
plt.title("Result of Hough Transform")
plt.axis('off')
```
